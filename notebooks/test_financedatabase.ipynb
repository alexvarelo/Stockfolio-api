```python
# Import required libraries
import yfinance as yf
import pandas as pd
from pprint import pprint

# 1. Get all available market data
all_markets = yf.Market
print("Available market data:")
print([attr for attr in dir(yf.market) if not attr.startswith('_')])

# 2. Get pre-market movers
print("\nPre-market movers:")
pre_market = yf.Market.pre_market
print(pre_market)

# 3. Get after-hours movers
print("\nAfter-hours movers:")
after_hours = yf.market.after_hours
print(after_hours)

# 4. Get most active stocks
print("\nMost active stocks (by volume):")
most_active = yf.market.most_active
print(most_active)

# 5. Get gainers
print("\nTop gainers:")
gainers = yf.market.gainers
print(gainers)

# 6. Get losers
print("\nTop losers:")
losers = yf.market.losers
print(losers)

# 7. Get market status
print("\nMarket status:")
market_status = yf.market.market_status
print(f"Market is currently: {'Open' if market_status == 'OPEN' else 'Closed'}")

# 8. Get market calendar (next market open/close times)
print("\nMarket calendar:")
market_calendar = yf.market.calendar
print(market_calendar)

# 9. Get market news
print("\nMarket news:")
market_news = yf.market.news
for i, article in enumerate(market_news[:5], 1):  # Show first 5 news items
    print(f"\n{i}. {article['title']}")
    print(f"   Source: {article.get('publisher', 'N/A')}")
    print(f"   Published: {article.get('publishedAt', 'N/A')}")

# 10. Get market summary
print("\nMarket summary:")
market_summary = yf.market.summary
print(market_summary)

# 11. Get market indices
print("\nMajor market indices:")
indices = yf.market.indices
print(indices)

# 12. Get cryptocurrency data
print("\nTop cryptocurrencies:")
crypto = yf.market.crypto
print(crypto)

# 13. Get market movers by index
print("\nDow Jones market movers:")
dow_movers = yf.market.movers('^DJI')
print(dow_movers)

# 14. Get market sectors performance
print("\nMarket sectors performance:")
sectors = yf.market.sectors
print(sectors)

# 15. Get market quotes for specific symbols
print("\nQuotes for specific symbols:")
quotes = yf.market.quotes(['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'META'])
print(quotes)
```
